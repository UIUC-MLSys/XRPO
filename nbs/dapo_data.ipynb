```python
from datasets import load_dataset

df = load_dataset("BytedTsinghua-SIA/DAPO-Math-17k")["train"].to_pandas()
df["index"] = df["extra_info"].apply(lambda x: x["index"])
df = df.drop_duplicates(subset="index", ignore_index=True)
```

```python
def fix_prompt(x):
    content = x[0]["content"]
    content = content.replace('Remember to put your answer on its own line after "Answer:".', "Let's think step by step and output the final answer within \\boxed{}.")
    content = content.replace("Solve the following math problem step by step. The last line of your response should be of the form Answer: $Answer (without quotes) where $Answer is the answer to the problem.\n\n", "")
    return [{'content': content, 'role': 'user'}]

df["prompt"] = df["prompt"].apply(fix_prompt)
```

```python
def fix_reward_model(x):
    x["style"] = "rule"
    return x

df["reward_model"] = df["reward_model"].apply(fix_reward_model)
```

```python
df.to_parquet("data/train_DAPO-Math-17k.parquet", index=False)
```
